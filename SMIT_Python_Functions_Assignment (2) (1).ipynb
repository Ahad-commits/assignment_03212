## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.

  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
# Write your code below
def calculate_salary(basic, allowance_percent, deduction_percent):

    def calculate_allowance():
        return basic * allowance_percent / 100

    def calculate_deduction():
        return basic * deduction_percent / 100

    def calculate_final_salary(allowance, deduction):
        return basic + allowance - deduction

    allowance = calculate_allowance()
    deduction = calculate_deduction()
    return calculate_final_salary(allowance, deduction)

final = calculate_salary(50000, 20, 10)
print(f"Final Salary: Rs. {final}")
````

````python
# Write your code below
def student_info(**details):
    return details

def get_total(marks):
    total = 0
    for m in marks:
        total += m
    return total

def get_percentage(total, subjects, max_per_subject=100):
    return total / (subjects * max_per_subject) * 100

def get_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    return "F"

is_pass = lambda percentage: "Pass" if percentage >= 40 else "Fail"

def result_card(info, subjects, total, percentage, grade, status):
    print("=" * 42)
    print("            RESULT CARD")
    print("=" * 42)
    for key, value in info.items():
        print(f"{key.capitalize():<12}: {value}")
    print("-" * 42)
    for subject, mark in subjects.items():
        print(f"{subject:<15}: {mark}")
    print("-" * 42)
    print(f"Total       : {total} / {len(subjects) * 100}")
    print(f"Percentage  : {percentage:.2f}%")
    print(f"Grade       : {grade}")
    print(f"Status      : {status}")
    print("=" * 42)

info = student_info(name="Sara Ahmed", roll_no="SMIT-045", course="Python", institute="SMIT")
subjects = {"English": 78, "Maths": 85, "Science": 72, "Urdu": 90, "Computer": 95}

total = get_total(list(subjects.values()))
percentage = get_percentage(total, len(subjects))
grade = get_grade(percentage)
status = is_pass(percentage)

result_card(info, subjects, total, percentage, grade, status)
````

````python
# Write your code below
# Product catalogue (dictionary)
catalog = {
    "laptop":     {"name": "Laptop",     "price": 85000},
    "mouse":      {"name": "Mouse",      "price": 1200},
    "keyboard":   {"name": "Keyboard",   "price": 2500},
    "headphones": {"name": "Headphones", "price": 4000},
}

def customer_info(name, city, phone):
    return {"name": name, "city": city, "phone": phone}

def product_details(product_key):
    return catalog[product_key]

def line_total(price, quantity):
    return price * quantity

def calculate_subtotal(*line_totals):                 # *args
    return sum(line_totals)

def calculate_discount(subtotal, rate=None):
    if rate is None:                                   # default: slab-based
        rate = 10 if subtotal >= 50000 else 5 if subtotal >= 10000 else 0
    return subtotal * rate / 100

def calculate_delivery(subtotal, city):
    if subtotal >= 20000:
        return 0                                       # free delivery
    return 200 if city.lower() == "karachi" else 400

def final_bill(subtotal, discount, tax_rate, delivery):
    # nested function for tax
    def tax_amount():
        return (subtotal - discount) * tax_rate / 100
    tax = tax_amount()
    return tax, subtotal - discount + tax + delivery

format_money = lambda x: f"Rs. {x:,.2f}"               # lambda

def place_order(customer, cart, tax_rate=16):
    lines, totals = [], []
    for key, qty in cart:                              # loop
        p = product_details(key)
        total = line_total(p["price"], qty)
        lines.append((p["name"], p["price"], qty, total))
        totals.append(total)

    subtotal = calculate_subtotal(*totals)
    discount = calculate_discount(subtotal)
    delivery = calculate_delivery(subtotal, customer["city"])
    tax, final = final_bill(subtotal, discount, tax_rate, delivery)
    order_summary(customer, lines, subtotal, discount, tax, tax_rate, delivery, final)

def order_summary(customer, lines, subtotal, discount, tax, tax_rate, delivery, final):
    print("=" * 56)
    print("                ORDER SUMMARY")
    print("=" * 56)
    print(f"Customer : {customer['name']}")
    print(f"City     : {customer['city']}")
    print(f"Phone    : {customer['phone']}")
    print("-" * 56)
    print(f"{'Product':<14}{'Price':>12}{'Qty':>6}{'Total':>18}")
    print("-" * 56)
    for name, price, qty, total in lines:
        print(f"{name:<14}{price:>12,}{qty:>6}{total:>18,}")
    print("-" * 56)
    print(f"{'Subtotal':<30}{format_money(subtotal):>26}")
    print(f"{'Discount':<30}{'-' + format_money(discount):>26}")
    print(f"{'Tax (' + str(tax_rate) + '%)':<30}{format_money(tax):>26}")
    print(f"{'Delivery':<30}{format_money(delivery):>26}")
    print("=" * 56)
    print(f"{'FINAL BILL':<30}{format_money(final):>26}")
    print("=" * 56)

customer = customer_info("Ali Khan", "Karachi", "0300-1234567")
cart = [("laptop", 1), ("mouse", 2), ("keyboard", 1)]
place_order(customer, cart)
````
